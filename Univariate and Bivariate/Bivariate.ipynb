```python
def QuanQual(dataset):

    Qual = []
    Quan = []
    for columnName in dataset.columns:
        if(dataset[columnName].dtype == 'O'):
           Qual.append(columnName)
        else:
           Quan.append(columnName)
    return Quan,Qual
```

```python
# Multicollinearity & VIF
```

```python
# Multicollinearity occurs when two or more independent/input variables are highly correlated with each other.

# VIF (Variance Inflation Factor) measures how much the variance of a regression coefficient is increased because of multicollinearity.

# VIF = 1 --> nothing, 1-5 ---> acceptable, >5 ---> high multicollinear, >10 ---> very high/ serious
```

```python
dataset[quan].columns
```

```python
# we import "variance_inflation_factor" from statsmodels lib
# this handles the calculation of VIF 
from statsmodels.stats.outliers_influence import variance_inflation_factor

# we drop 'meal_id' because it dosen't do any significant contribution.
# we drop 'is_healthy' because it is the target
X = dataset.drop(['meal_id', 'is_healthy'], axis=1)

# function is defined with an argument "X"
# X is considered as the dataframe that holds the numarical coloum/feature fro which u need to calculate VIF 
def calc_vif(X):
    
    # create a empty variable "vif" to hold the o/p
    vif = pd.DataFrame()
    
    # create a coloum for output variable 
    # "X.columns" pairs each coloumn name directely with its calculated VIF score in the o/p table 
    vif["variables"] = X.columns
    
    # create a seacond coloum name "VIF"
    # x.shape[1] ----> get the total no of columns in X
    # range(X.shape[1]) ----> loops through column indices
    # X.values ----> converts the dataframe into a raw array
    # variance_inflation_factor(X.values, i) ----> calculate the VIF for the ith column
    vif["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
    return(vif)
```

```python
calc_vif(dataset[["calories","protein_g","carbs_g","fat_g","fiber_g","sugar_g","sodium_mg","cholesterol_mg","rating"]])
```

```python
# Hypothesis & T testing 
```

```python
# Hypothesis testing is a statistical method used to determine whether there is enough evidence to accept or reject a claim about a population.

# Two hypotheses:
# H₀ (Null Hypothesis): No significant difference/relationship exists.
# H₁ (Alternative Hypothesis): A significant difference/relationship exists.

# Using p-value:
# p-value < 0.05 → Reject H₀ → Significant
# p-value ≥ 0.05 → Fail to reject H₀ → Not significant
```

```python
# A t-test is a statistical test used to determine whether there is a significant difference between the means of two groups.
```

```python
from scipy.stats import ttest_ind

healthy = dataset[dataset['is_healthy'] == 1]['calories']
unhealthy = dataset[dataset['is_healthy'] == 0]['calories']

ttest_ind(healthy, unhealthy)
```

```python
# Since the p value(8.6444) is > 0.05 we accept null hypothesis and reject alternative hypothesis
```
